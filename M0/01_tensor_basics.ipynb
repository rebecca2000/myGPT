# M0 · 01 — Tensor basics

A **tensor** is just an n-dimensional array of numbers (like a NumPy array) that
PyTorch can run fast and differentiate. Scalars, vectors, matrices, and the
`(batch, time, channel)` blocks in the GPT are *all* tensors — they only differ
in **how many dimensions** they have.

## 2. Dimensions (rank): 0-D, 1-D, 2-D

The number of axes is the tensor's **rank** (`.ndim`).
- scalar `torch.tensor(5)` → 0-D
- vector `[5]` → 1-D
- matrix `[[5]]` → 2-D

## ✅ Recap

- A tensor = n-dimensional array; **rank** = number of axes (`.ndim`).
- `.shape` = size per axis; you'll track `(B, T, C)` with it.
